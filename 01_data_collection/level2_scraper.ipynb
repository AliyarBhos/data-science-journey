```python
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://www.ifixit.com/Device/Acer_Aspire_1_A114-31'
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) Chrome/124.0.0.0'}

response = requests.get(url, headers=headers, timeout=10)
html = response.text
```

```python
soup = BeautifulSoup(html, 'html.parser')
```

```python
title_tag = soup.find('title')
print(title_tag)
print(title_tag.text)
print(title_tag.text.strip())
```

### Let's find things by class/attribute

```python
first_link = soup.find('a', class_='skip-to-content')

all_links = soup.find_all('a', class_='skip-to-content')
for link in all_links:
    print(link.text.strip(), '->', link.get('href'))
```
