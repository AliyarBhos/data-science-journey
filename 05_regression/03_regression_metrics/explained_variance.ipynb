### Explained Variance (R²)

- Measures how much of the variation in the target variable (\(y\)) is explained by the regression model.
- Ranges from \(-\infty\) to \(1\).
- \(R^2 = 1\): Perfect fit.
- \(R^2 = 0\): Model performs like predicting the mean.
- \(R^2 < 0\): Model performs worse than predicting the mean.

**Formula:**

$$
R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
$$

- \(SS_{res}\): Unexplained variation (squared prediction errors).
- \(SS_{tot}\): Total variation around the actual mean.

**Note:** A high \(R^2\) does not necessarily mean a good model. Evaluate it alongside RMSE and MAE.

### Adjusted R²

- Adjusts R² based on the number of features and observations.
- Penalizes unnecessary features.
- Increases only when a new feature improves the model sufficiently.
- Useful for comparing regression models with different numbers of features.

**Formula:**

$$
R^2_{adj} = 1 - (1-R^2)\frac{n-1}{n-p-1}
$$

Where:
- \(n\): Number of observations.
- \(p\): Number of independent variables (features).
- \(R^2\): Regular coefficient of determination.
