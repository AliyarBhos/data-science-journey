```python
import numpy as np
from sklearn.metrics import mean_absolute_percentage_error
```

```python
y_true = np.array([100, 200, 300])
y_pred = np.array([90, 220, 240])
```

```python
mape = mean_absolute_percentage_error(y_true, y_pred) * 100
denominator = np.abs(y_true) + np.abs(y_pred)
```

```python
smape_terms = np.divide(2 * np.abs(y_true - y_pred), denominator, out=np.zeros_like(denominator, dtype=float), where=denominator != 0)
```
