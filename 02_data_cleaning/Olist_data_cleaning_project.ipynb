##### Now I need to determine the reasons behind missing values

```python
df5.groupby(df5['review_comment_title'].isnull())['review_score'].mean()
```

```python
df5.groupby(df5['review_comment_message'].isnull())['review_score'].mean()
```

##### Both cases above are MAR. missingness depends on another observed column (review_score), not on the hidden content of the comment itself.

```python
both = df5[df5['review_comment_title'].notnull() & df5['review_comment_message'].notnull()].shape[0]
only_message = df5[df5['review_comment_title'].isnull() & df5['review_comment_message'].notnull()].shape[0]
only_title = df5[df5['review_comment_title'].notnull() & df5['review_comment_message'].isnull()].shape[0]
neither = df5[df5['review_comment_title'].isnull() & df5['review_comment_message'].isnull()].shape[0]

print(f'How many people fill both title and message -> {both}')
print(f'How many people fill just a message -> {only_message}')
print(f'How many people fill just a title -> {only_title}')
print(f'How many people skip both entirely (never opened the comment box at all) -> {neither}')
```
